```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/Users/ochsneto/Documents/BDMM/BDMM-Flow/metrics.csv")
```

```python
df.groupby("strategy").mean()
```

```python
by_time = df.groupby(["time", "strategy"]).mean()
```

```python
sns.lineplot(by_time, x="time", y="condition", hue="strategy")
plt.yscale("log")
t0 = 7500
plt.xlim(t0, t0 + 600)
plt.ylim(0, 10000000000000)
```
