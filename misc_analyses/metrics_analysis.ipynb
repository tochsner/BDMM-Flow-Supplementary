```python
import pandas as pd
import seaborn as sns                                                                                     
import matplotlib.pyplot as plt
```

```python
sns.set_theme(style="darkgrid")
```

```python
results = pd.read_csv("/Users/ochsneto/Documents/BDMM/BDMM-Flow/metrics.csv")
results.describe()
```

```python
sns.lineplot(
    data=results[results["metric"] == "flow_norm"],
    x="time",
    y="value",
    hue="algorithm",
    errorbar=("pi", 50),
    estimator="median"
)
```

```python
sns.lineplot(
    data=results[results["metric"] == "jacobian_norm"],
    x="time",
    y="value",
    hue="algorithm",
    errorbar=("pi", 50),
    estimator="median"
)
plt.ylabel("||AY||")
```

```python
sns.lineplot(
    data=results[results["metric"] == "normalized_jacobian"],
    x="time",
    y="value",
    hue="algorithm",
    errorbar=("pi", 50),
    estimator="median"
)
plt.ylabel("||AY|| / ||Y||")
```

```python
sns.lineplot(
    data=results[results["metric"] == "jacobian_min_ev"],
    x="time",
    y="value",
    hue="algorithm",
    errorbar=("pi", 50),
    estimator="median"
)
plt.ylabel("min(eig(J))")
```

```python
sns.lineplot(
    data=results[results["metric"] == "jacobian_max_ev"],
    x="time",
    y="value",
    hue="algorithm",
    errorbar=("pi", 50),
    estimator="median"
)
plt.ylabel("max(eig(J))")
```

```python
sns.lineplot(
    data=results[results["metric"] == "jacobian_ev_ratio"],
    x="time",
    y="value",
    hue="algorithm",
    errorbar=("pi", 50),
    estimator="median"
)
plt.ylabel("min(eig(J)) / max(eig(J))")
plt.yscale("log")
```

```python
sns.lineplot(
    data=results[results["metric"] == "system_bound"],
    x="time",
    y="value",
    hue="algorithm",
    errorbar=("pi", 50),
    estimator="median"
)
plt.ylabel("min(eig(J)) / max(eig(J))")
```
