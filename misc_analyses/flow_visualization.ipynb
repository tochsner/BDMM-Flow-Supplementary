```python
import numpy as np
import matplotlib.pyplot as plt

plt.xkcd()
```

```python
# recreate plot from screenshot
# define time points
t_0 = 0.0
t_1 = 2.0

# blue trajectory [0.0 -> 2.0] starting at 1.0
t_blue = np.linspace(t_0, t_1, 100)
x_blue_start = 1.0
A_blue = (x_blue_start - np.sin(t_0)) / np.exp(-t_0)
y_blue = A_blue * np.exp(-t_blue) + np.sin(t_blue)

# red trajectory [0.25 -> 1.5] starting at 1.1
t_red_start = 0.25
t_red_end = 1.5
x_red_start = 1.1
t_red = np.linspace(t_red_start, t_red_end, 100)
A_red = (x_red_start - np.sin(t_red_start)) / np.exp(-t_red_start)
y_red = A_red * np.exp(-t_red) + np.sin(t_red)

# create the plot
plt.figure(figsize=(10, 6))
plt.plot(t_blue, y_blue, 'b-', linewidth=2, label=f'[{t_0} -> {t_1}] starting at {x_blue_start}')
plt.plot(t_red, y_red, 'r-', linewidth=2, label=f'[{t_red_start} -> {t_red_end}] starting at {x_red_start}')

# add vertical dashed lines for red trajectory
plt.axvline(x=t_red_start, color='lightcoral', linestyle='--', alpha=0.5)
plt.axvline(x=t_red_end, color='lightcoral', linestyle='--', alpha=0.5)

# add horizontal dashed line at starting value of red trajectory
plt.axhline(y=x_red_start, color='lightcoral', linestyle='--', alpha=0.5)

# labels and formatting
plt.xlabel('Time (t)', fontsize=12)
plt.ylabel('State (x)', fontsize=12)
plt.legend()
plt.ylim([0.975, 1.275])
plt.tight_layout()
plt.show()
```
