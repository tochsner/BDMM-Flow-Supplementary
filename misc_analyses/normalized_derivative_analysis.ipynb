```python
from scipy.integrate import solve_ivp
from scipy import linalg
import numpy as np
import matplotlib.pyplot as plt
```

```python
dim = 5
```

```python
for _ in range(1000):
    A = np.random.normal(0, 2, size=(dim, dim))

    norms_a = []
    norms_b = []

    for _ in range(10_000):
        X = np.random.uniform(0, 1, size=(dim, dim))
        I = np.identity(dim)

        norms_a.append(np.linalg.norm(A @ I) / np.linalg.norm(I))
        norms_b.append(np.linalg.norm(A @ X) / np.linalg.norm(X))

    assert np.mean(norms_a) > np.mean(norms_b)

    print("OK")
```

```python
np.mean(norms_a), np.mean(norms_b)
```
